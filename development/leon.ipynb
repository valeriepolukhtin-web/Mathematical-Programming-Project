```python
import nltk
nltk.download('gutenberg')
from nltk.corpus import gutenberg
moby = gutenberg.raw('melville-moby_dick.txt')
print(moby[:200])
```

```python
import string
lowercase = list(string.ascii_lowercase)
uppercase = list(string.ascii_uppercase)
def q2cipher(x):
    list1 = list(x)
    list2 = []
    for i in list1:
        if i in lowercase or i in uppercase:
            list2.append(i.upper())
        else:
            list2.append(' ')
    updatedtext = ' '.join(''.join(list2).split())
    return updatedtext
```

```python
newmoby = q2cipher(moby)
newmoby[:200]
```
