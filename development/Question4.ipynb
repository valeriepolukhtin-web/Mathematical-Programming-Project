```python
messages = [ciphertexts[2], ciphertexts[3], ciphertexts[4]]
decryptedMessages = []
for message in messages:
    charFreq_Cipher = dict()
    for i in range(len(message)):
        currentChar = message[i]
        if (currentChar in charFreq_Cipher):
            charFreq_Cipher[currentChar] += 1
        else:
            charFreq_Cipher[currentChar] = 1
        
    sortedCipher = sorted(charFreq_Cipher.items(), key=lambda x: x[1], reverse=True)
    cipherToEnglish = {}

    for i in range(len(sortedCipher)):
        cipherChar = sortedCipher[i][0]
        englishChar = sortedEnglish[i][0]

        cipherToEnglish[cipherChar] = englishChar

    decryptedMessage = ""
    for char in message:
        decryptedMessage += cipherToEnglish[char]
    print(f"Ciphertext:")
    print(decryptedMessage)
    decryptedMessages.append(decryptedMessage) #Adding to list so we can isolate each message to decrypt it 
    print(f"Character shift:")
    for cipherChar, englishChar in cipherToEnglish.items():
        print(f"{cipherChar} -> {englishChar}")
```

### Now implimenting part 4's swap function:

```python
#Code to swap any two characters 'i' and 'j'.
#The code replaces one character with another rather than swapping two characters simultaneously.
#A placeholder is necessary so the newly replaced characters aren't replaced again during latr replacements.
def swaproles(s,i,j):
    return(s.replace(i,'-').replace(j,i).replace('-',j))
```
