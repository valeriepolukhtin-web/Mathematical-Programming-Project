```python
def calcFreq_i (i, ciphertext, charFreq_Cipher):
    if (i in charFreq_Cipher):
        charFreq_Cipher[i] += 1
    else:
        charFreq_Cipher[i] = 1
    return charFreq_Cipher[i]
```

```python
def calcFreq_ij (ij, ciphertext, strFreq_Cipher):
    if (ij in strFreq_Cipher):
        strFreq_Cipher[ij] += 1
    else:
        strFreq_Cipher[ij] = 1
    return strFreq_Cipher[ij]
```

```python
def prob(text):
    bigram_prob = dict()
    charFreq_Cipher = dict()
    strFreq_Cipher = dict()

    for i in range(len(text)):
        if text[i] in charFreq_Cipher:
            charFreq_Cipher[text[i]] += 1
        else:
            charFreq_Cipher[text[i]] = 1

    for i in range(len(text) - 1):
        ij = text[i] + text[i+1]

        if ij in strFreq_Cipher:
            strFreq_Cipher[ij] += 1
        else:
            strFreq_Cipher[ij] = 1

    alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ "

    for i in alphabet:
        for j in alphabet:
            ij = i + j

            freq_ij = strFreq_Cipher.get(ij, 0)
            freq_i = charFreq_Cipher.get(i, 0)

            probability = (freq_ij + 1) / (freq_i + 27)
            bigram_prob[ij] = probability

    return bigram_prob
```

```python
def score(m):
    S=0
    probabilities = prob(newmoby)
    for i in range(len(m) - 1):
        ij = m[i] + m[i+1]
        S = S + math.log(probabilities.get(ij, 0))
    return S
```
