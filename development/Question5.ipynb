```python
import math
import random
```

```python
def calcFreq_i (i, ciphertext, charFreq_Cipher):
    if (i in charFreq_Cipher):
        charFreq_Cipher[i] += 1
    else:
        charFreq_Cipher[i] = 1
    return charFreq_Cipher[i]
```

```python
def calcFreq_ij (ij, ciphertext, strFreq_Cipher):
    if (ij in strFreq_Cipher):
        strFreq_Cipher[ij] += 1
    else:
        strFreq_Cipher[ij] = 1
    return strFreq_Cipher[ij]
```

```python
def prob(text):
    bigram_prob = dict()
    charFreq_Cipher = dict()
    strFreq_Cipher = dict()
    for i in range(len(text) - 1):
        freq_i = calcFreq_i(text[i], text, charFreq_Cipher)
        ij = text[i] + text[i+1]
        freq_ij = calcFreq_ij(ij, text, strFreq_Cipher)
        probability = (freq_ij + 1) / freq_i
        if (ij in bigram_prob):
            bigram_prob[ij] += 1
        else:
            bigram_prob[ij] = 1
    return bigram_prob
```

```python
def score(m):
    S=0
    probabilities = prob(m)
    probValues_list = list(probabilities.values())
    for index in range(len(probabilities)):
        S = S + math.log(probValues_list[index])
    return S
```
