```python
with open("../data/newmoby.txt", "r", encoding="utf-8") as file:
    newmoby = file.read()
```

```python
charFreq_NewMoby = dict()

for i in range(len(newmoby)):
    currentChar = newmoby[i]
    if (currentChar in charFreq_NewMoby):
        charFreq_NewMoby[currentChar] += 1
    else:
        charFreq_NewMoby[currentChar] = 1
        
sortedEnglish = sorted(charFreq_NewMoby.items(), key=lambda x: x[1], reverse=True)
```

```python
charFreq_Cipher = dict()

# Choosing to decrypt the second cipher first as it is the longest
ciphertext = ciphertexts[2]

for i in range(len(ciphertext)):
    currentChar = ciphertext[i]
    if (currentChar in charFreq_Cipher):
        charFreq_Cipher[currentChar] += 1
    else:
        charFreq_Cipher[currentChar] = 1
        
sortedCipher = sorted(charFreq_Cipher.items(), key=lambda x: x[1], reverse=True)
```

```python
cipherToEnglish = {}

for i in range(len(sortedCipher)):
    cipherChar = sortedCipher[i][0]
    englishChar = sortedEnglish[i][0]

    cipherToEnglish[cipherChar] = englishChar

decryptedMessage = ""

for char in ciphertext:
    decryptedMessage += cipherToEnglish[char]

print(decryptedMessage)

for cipherChar, englishChar in cipherToEnglish.items():
    print(f"{cipherChar} -> {englishChar}")
```

### Now implimenting part 4's swap function:

```python
# Khianna's character swap function
def swaproles(s,i,j):
    return(s.replace(i,'-').replace(j,i).replace('-',j))
```

```python
# Keep a copy of the result from part 3
current_message = decryptedMessage
print(current_message)
```

```python
# Choosing to swap 'S' with 'H' to create 'THE' from 'TSE'
trial1_message = swaproles(current_message, 'S', 'H')
print(trial1_message)
```

```python
# Carry on from trial 1 as it seemed a good guess. Now swap 'I' and 'S' to reveal 'HOLMES' from 'HOLMEI'
trial2_message = swaproles(trial1_message, 'I', 'S')
print(trial2_message)
```

```python
# Swap 'V' and 'K' to reveal 'SHERLOCK' from 'SHERLOCV'
trial3_message = swaproles(trial2_message, 'V', 'K')
print(trial3_message)
```

```python
# Using intuitive guesses in the opening phrase to choose further swaps
# 'G' to 'P': 'GRESERVED' to 'PRESERVED'
# 'N' to 'I': 'HNS' to 'HIS'
# 'W' to 'F': 'PROWESSIONAL' to 'PROFESSIONAL'
trial4_message = swaproles(trial3_message, 'G', 'P')
trial4_message = swaproles(trial4_message, 'N', 'I')
trial4_message = swaproles(trial4_message, 'W', 'F')
print(trial4_message)
```

```python
# Swap 'W' and 'U' to reveal 'UNTIL OUR' from 'WNTIL OWR'
trial5_message = swaproles(trial4_message, 'W', 'U')
print(trial5_message)
```

```python
# Swap 'B' for 'Y' to reveal:
# 'EASY' from 'EASB'
# 'PROFOUNDLY' from 'PROFOUNDLB'
trial6_message = swaproles(trial5_message, 'B', 'Y')
print(trial6_message)
```

```python
# Swap 'Q' for 'X' to reveal 'EXCITED' from 'EQCITED' to complete the first sentences
trial7_message = swaproles(trial6_message, 'Q', 'X')
print(trial7_message)
```

```python
# Swap 'J' for 'Z' to reveal 'PUZZLED' from 'PUJJLED' to completely decipher the text.
final_message = swaproles(trial7_message, 'J', 'Z')
print(final_message)
```
