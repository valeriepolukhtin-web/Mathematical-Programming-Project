```python
import string
# Creating a string, in order, of all uppercase letters and space
letters = string.ascii_uppercase + " "

# Defining a function to decode a message with a shift of n characters
def cyclic_shift(message, n): 
    # Ensuring message is in upper case letters
    message = message.upper()
    # Creating an empty string for the decrypted message
    decrypt_message = ""
    for char in message: 
        # Finding index of current character within string containing all possible char
        m = letters.index(char)
        # We want to shift our characters back by n (the amount of cyclic shifting in the original encryption)
        m = m - n
        # Ensuring our index has not become negative
        m = m % len(letters)
        # Adding appropriate char corresponding to index to our decrypted string
        decrypt_message += letters[m]
    return decrypt_message
```
