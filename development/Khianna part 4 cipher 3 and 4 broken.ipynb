```python
charFreq_NewMoby = dict()

for i in range(len(newmoby)):
    currentChar = newmoby[i]
    if (currentChar in charFreq_NewMoby):
        charFreq_NewMoby[currentChar] += 1
    else:
        charFreq_NewMoby[currentChar] = 1
        
sortedEnglish = sorted(charFreq_NewMoby.items(), key=lambda x: x[1], reverse=True)
```

```python
def swaproles(s,i,j):
    return(s.replace(i,'-').replace(j,i).replace('-',j))
```
