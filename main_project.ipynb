## Question 1 (Core): 
In the first message, the cipher is simply a cyclic shift of the 27 characters. E.g. Considering the ordered alphabet, a shift of 2 rightwards would map A → C, B → D, ..., Y → space, Z → A, space → B. Find the shift and decode the message. One simple approach is to try all possibilities and check by hand which one produces readable text.

We will start by writing a function that takes a message and a shift and outputs the message in which all characters are shifted that much back to the left. 

```python
# Creating a string, in order, of all uppercase letters and space
letters = string.ascii_uppercase + " "

# Defining a function to decode a message with a shift of n characters
def cyclic_shift(message, n): 
    # Ensuring message is in upper case letters
    message = message.upper()
    # Creating an empty string for the decrypted message
    decrypt_message = ""
    for char in message: 
        # Finding index of current character within string containing all possible char
        m = letters.index(char)
        # We want to shift our characters back by n (the amount of cyclic shifting in the original encryption)
        m = m - n
        # Ensuring our index has not become negative
        m = m % len(letters)
        # Adding appropriate char corresponding to index to our decrypted string
        decrypt_message += letters[m]
    return decrypt_message
```

```python
# Import NLTK, which provides access to a collection of text
import nltk

# Download the Gutenberg corpus of literary texts
nltk.download('gutenberg')

# Import the Gutenberg corpus and extract the raw text of Moby Dick
from nltk.corpus import gutenberg
moby = gutenberg.raw('melville-moby_dick.txt')

# Display the first 200 characters to check that the text was loaded
print(moby[:200])

# Create lists containing the lowercase and uppercase English alphabet
lowercase = list(string.ascii_lowercase)
uppercase = list(string.ascii_uppercase)
```

```python
# Convert the text into the 27-character alphabet used by the cipher
def q2cipher(x):
    # Convert the input string into a list of individual characters
    list1 = list(x)
    list2 = []
    
    # Keep letters and convert them to uppercase
    # Replace all other characters with spaces
    for i in list1:
        if i in lowercase or i in uppercase:
            list2.append(i.upper())
        else:
            list2.append(' ')

    # Collapse consecutive spaces and remove spaces at the beginning and end of the text
    updatedtext = ' '.join(''.join(list2).split())
    
    # Return the cleaned text
    return updatedtext
```

```python
# Apply the cleaning function to the Moby Dick sample
newmoby = q2cipher(moby)
# Display the first 200 characters of the cleaned text
newmoby[:200]
```

## Question 5 (Core): 

The above method is less likely to work for the later messages, because they are not
long enough to get accurate frequency counts (and some might involve unusual language).
Instead use the following approach based on ideas from statistical physics. In the sample text, record the frequency (i.e. number of occurrences) of each of the $27^2$ bigrams, i.e. pairs of consecutive characters such as SH, as well as the $27$ characters. For each pair of characters $i, j$ compute $$p(i, j) := \frac{\text{frequency}(ij) + 1}{\text{frequency}(i)}.$$ This represents the probability that $i$ is followed by $j$; the $+1$ is a ‘fudge factor’ to avoid zeros. To any potential decoded message $m = i_1i_2 . . . i_n$ we assign a score $$S(m) = \sum_{j=1}^{n-1} \text{log}p(i_j,i_{j+1})$$ which measures how plausible $m$ is as a piece of English text (it represents the logarithm of the probability of seeing $m$ under a certain “Markov model”). Write a function that computes $S(m)$ for any message $m$.

We start by defining a function using the cleaned Moby-Dick text that will create a probability table of all the $27^2$ possible bigrams.

```python
def prob(text):
    # Setting up our dictionaries: bigram probabilities, character counts and bigram counts
    bigram_prob = dict()
    charFreq_Cipher = dict()
    strFreq_Cipher = dict()

    # Counting the amount of each character that starts a bigram in the reference text
    for i in range(len(text) - 1):  # We dont include the last character as it has no following one to make a bigram with
        if text[i] in charFreq_Cipher:
            charFreq_Cipher[text[i]] += 1
        else:
            charFreq_Cipher[text[i]] = 1

    # Counting how often each overlapping bigram occurrs by pairing each charcter with the following one 
    for i in range(len(text) - 1):
        ij = text[i] + text[i+1]

        if ij in strFreq_Cipher:
            strFreq_Cipher[ij] += 1
        else:
            strFreq_Cipher[ij] = 1

    
    alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ "

    # Building a probability table for every possible bigram including those not found in the cleaned Moby-Dick text
    for i in alphabet:
        for j in alphabet:
            ij = i + j

            # Creating variables that lookup the counts for a particular bigram and its starting character 
            # If either key is missing use 0
            freq_ij = strFreq_Cipher.get(ij, 0)
            freq_i = charFreq_Cipher.get(i, 0)

            # Implimenting the formula with +1 to the numerator, as in the breif, to ensure we can take the score of unseen bigrams
            # Adding 27 to the denominator to account for one extra count per possible characters that could follow "i"
            probability = (freq_ij + 1) / (freq_i + 27)  
            bigram_prob[ij] = probability

    return bigram_prob
```

Now we can build the probability table and use it when defining our score function.

```python
# Building the probability table of the cleaned Moby-Dick text once and assigning it to a variable
probability_newmoby = prob(newmoby)

def score(m):
    S=0
    # Looping through each bigram in message m
    # We stop before the last character as it would have no partner character
    for i in range(len(m) - 1):
        ij = m[i] + m[i+1]
        # Looking up the probability of the bigram in the Moby-Dick table
        # Adding its log(p) to the running score for every occurance in m
        S = S + math.log(probability_newmoby[ij])
    return S
```

Choose two characters at random and swap their roles in the cipher, to get a new candidate message $m'$. If $S(m') > S(m)$ then replace $m$ with $m'$. If $S(m') ≤ S(m)$ then replace $m$ with $m'$ with probability $$\text{exp}\frac{S(m')-S(m)}{T},$$ otherwise just keep m.

We now implement the Metropolis algorithm. Each call to metropolis_algo(m, T) proposes only one swap and returns either the new message or the previous, it then loops through the process using the returned guess as the starting point each time.

```python
all_char = list(string.ascii_uppercase) + [' ']

def metropolis_algo(m, T): 
    if T <= 0:
        return "Error: T is not greater than 0."
    # Keeping the old message
    old_message = m
    # Assigning two random characters to i and j
    i = all_char[random.randint(0, len(all_char) - 1)]
    j = all_char[random.randint(0, len(all_char) - 1)]
    # Checking i and j are not the same character
    while i == j: 
        j = all_char[random.randint(0, len(all_char) - 1)]

    # Swapping i and j with our function from part 4
    new_message = swaproles(m,i,j)

    # If score of new message greater than score of old message, then we replace
    if score(new_message) > score(old_message):
        return new_message
    # For scores equal or lower, we use the exponential expression from the brief 
    # This calculates the chance of accepting the swap
    prob = math.e**((score(new_message) - score(old_message)) / T)
    # If randomly generated number b/w 0 and 1 is below our probability, replace with new message
    if random.random() < prob:
        return new_message
    
    return old_message
```

```python
# Setting the seed to return consitent results
random.seed(987654321)

# We set the number of iterations to 100,000 and T = 1 to start
n = 100000
T = 1
m1 = ciphertexts[5]
print(f"Original message: \n{m1}, \n")

# Looping through n swap attemps, printing the score and current state of the message every 10,000 iterations
for i in range(n):
    m1 = metropolis_algo(m1,T)
    if (i + 1) % 10_000 == 0: 
        print(f"Iteration {i + 1:,} | score {score(m1):.2f}")
        print(m1, "\n")

print(f"Final result: {m1}")
```

#### **Simulated Annealing**

The breif suggests that we go from starting with T$= 10$ to T$= 0.1$, reducing it by a constant ratio each step. For $100,000$ iterations it has $99,999$ drops between each one. To work out our constant ratio ($r$) lets set up an equation $$10r^{99,999} = 0.1 .$$ Rearranging leaves us with $$r = (0.01)^{1/99,999}\approx0.99995 \text{(to }5\text{d.p)}.$$

```python
# We try again with ciphertexts[6] this time with varying T

# Choosing the same random seed to keep results consistent
random.seed(987654321)

# We keep the number of iterations at 100,000, but this time start with T = 10
n = 100000
T = 10
r = (0.1 / T) ** (1 / (n - 1)) # We re-write our value of r to work with differing values of T and iterations
m1 = ciphertexts[6]
print(f"Original message: \n{m1}, \n")

# Looping through n swap attemps, printing the current state of the message every 10,000 iterations, the score, and the value of T
for i in range(n):
    m1 = metropolis_algo(m1,T)
    if (i + 1) % 10_000 == 0: 
        print(f"Iteration {i + 1:,} | score {score(m1):.2f} | T = {T:.2f}")
        print(m1, "\n")
    # Multiplying by our constant ratio after each iteration 
    T = T * r
print(f"Final result: {m1}")
```
