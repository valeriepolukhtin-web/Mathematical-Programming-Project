## Question 1 (Core): 
In the first message, the cipher is simply a cyclic shift of the 27 characters. E.g. a shift of 2 would map A → C, B → D, ..., Y → space, Z → A, space → B. Find the shift and decode the message. One simple approach is to try all possibilities and check by hand which one produces readable text.

We will start by writing a function that takes a message and a shift and outputs the message in which all characters are shifted that much back to the left. 

```python
# Creating a string, in order, of all uppercase letters and space
letters = string.ascii_uppercase + " "

# Defining a function to decode a message with a shift of n characters
def cyclic_shift(message, n): 
    # Ensuring message is in upper case letters
    message = message.upper()
    # Creating an empty string for the decrypted message
    decrypt_message = ""
    for char in message: 
        # Finding index of current character within string containing all possible char
        m = letters.index(char)
        # We want to shift our characters back by n (the amount of cyclic shifting in the original encryption)
        m = m - n
        # Ensuring our index has not become negative
        m = m % len(letters)
        # Adding appropriate char corresponding to index to our decrypted string
        decrypt_message += letters[m]
    return decrypt_message
```

We start by writing a function that implements the Metropolis Algorithm. 

```python
all_char = list(string.ascii_uppercase) + [' ']

def metropolis_algo(m, T): 
    if T <= 0:
        return "Error: T is not greater than 0."
    # Keeping the old message
    old_message = m
    # Assigning two random characters to i and j
    i = all_char[random.randint(0, len(all_char) - 1)]
    j = all_char[random.randint(0, len(all_char) - 1)]
    # Checking i and j are not the same character
    while i == j: 
        j = all_char[random.randint(0, len(all_char) - 1)]

    # Swapping i and j
    new_message = swaproles(m,i,j)
    print(f"Score of old message: {score(old_message)} \nScore of new message: {score(new_message)}")
    # print(f"{i}, {j} \nOld message: {old_message} \nNew message: {new_message}")
    print(f"{i}, {j}, {new_message == old_message}")

    # If score of new message greater than score of old message, then we replace
    if score(new_message) > score(old_message):
        return new_message
    # Find the exponential expression value
    prob = math.e**((score(new_message) - score(old_message)) / T)
    # If randomly generated number b/w 0 and 1 is above our probability, will replace with new message
    if random.random() > prob:
        return new_message
    print("Not replaced.")
    return old_message
```

Above, I've written a test function for the metropolis algorithm that takes a message m and parameter T>0 and returns an updated message based on two randomly selected characters. Now, we will run this function multiple times. 

```python
n = 100
T = 1
m1 = ciphertexts[8]
print(f"Original message: \n{m1}")

for i in range(n):
    m1 = metropolis_algo(m1,T)
    if i % 1000 == 0: 
        print(f"Iteration {i}: {m1}")

print(f"Final result: {m1}")
```
