## Question 1 (Core): 
In the first message, the cipher is simply a cyclic shift of the 27 characters. E.g. Considering the ordered alphabet, a shift of 2 rightwards would map A → C, B → D, ..., Y → space, Z → A, space → B. Find the shift and decode the message. One simple approach is to try all possibilities and check by hand which one produces readable text.

We will start by writing a function that takes a message and a shift and outputs the message in which all characters are shifted that much back to the left. 

```python
# Creating a string, in order, of all uppercase letters and space
letters = string.ascii_uppercase + " "

# Defining a function to decode a message with a shift of n characters
def cyclic_shift(message, n): 
    # Ensuring message is in upper case letters
    message = message.upper()
    # Creating an empty string for the decrypted message
    decrypt_message = ""
    for char in message: 
        # Finding index of current character within string containing all possible char
        m = letters.index(char)
        # We want to shift our characters back by n (the amount of cyclic shifting in the original encryption)
        m = m - n
        # Ensuring our index has not become negative
        m = m % len(letters)
        # Adding appropriate char corresponding to index to our decrypted string
        decrypt_message += letters[m]
    return decrypt_message
```

```python
# Import NLTK, which provides access to a collection of text
import nltk

# Download the Gutenberg corpus of literary texts
nltk.download('gutenberg')

# Import the Gutenberg corpus and extract the raw text of Moby Dick
from nltk.corpus import gutenberg
moby = gutenberg.raw('melville-moby_dick.txt')

# Display the first 200 characters to check that the text was loaded
print(moby[:200])

# Create lists containing the lowercase and uppercase English alphabet
lowercase = list(string.ascii_lowercase)
uppercase = list(string.ascii_uppercase)
```

```python
# Convert the text into the 27-character alphabet used by the cipher
def q2cipher(x):
    # Convert the input string into a list of individual characters
    list1 = list(x)
    list2 = []
    
    # Keep letters and convert them to uppercase
    # Replace all other characters with spaces
    for i in list1:
        if i in lowercase or i in uppercase:
            list2.append(i.upper())
        else:
            list2.append(' ')

    # Collapse consecutive spaces and remove spaces at the beginning and end of the text
    updatedtext = ' '.join(''.join(list2).split())
    
    # Return the cleaned text
    return updatedtext
```

```python
# Apply the cleaning function to the Moby Dick sample
newmoby = q2cipher(moby)
# Display the first 200 characters of the cleaned text
newmoby[:200]
```

## Question 5 (Core): 

Question 5 asks the following. In the sample text, record the frequency (i.e. number of occurrences) of each of the 272 bigrams, i.e. pairs of consecutive characters such as SH, as well as the 27 characters. For each pair of characters i, j compute p(i, j) (defined). This represents the probability that i is followed by j; the +1 is a ‘fudge factor’ to avoid zeros. To any potential decoded message m = i1i2 . . . in we assign a score S(m) (defined) which measures how plausible m is as a piece of English text (it represents the logarithm of the probability of seeing m under a certain “Markov model”). Write a function that computes S(m) for any message m.

```python
# Calculate the probability of every possible bigram using the English corpus
def prob(text):
    bigram_prob = dict()
    charFreq_Cipher = dict()
    strFreq_Cipher = dict()

    # Count the frequency of each individual character
    for i in range(len(text)):
        if text[i] in charFreq_Cipher:
            charFreq_Cipher[text[i]] += 1
        else:
            charFreq_Cipher[text[i]] = 1

    # Count the frequency of each consecutive pair of characters
    for i in range(len(text) - 1):
        ij = text[i] + text[i+1]

        if ij in strFreq_Cipher:
            strFreq_Cipher[ij] += 1
        else:
            strFreq_Cipher[ij] = 1

    alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ "

    # Calculate probabilities for all 27 x 27 possible bigrams
    for i in alphabet:
        for j in alphabet:
            ij = i + j

            freq_ij = strFreq_Cipher.get(ij, 0)
            freq_i = charFreq_Cipher.get(i, 0)

            probability = (freq_ij + 1) / (freq_i + 27)
            bigram_prob[ij] = probability

    return bigram_prob
```

```python
# Calculate the score of a candidate plaintext using the English bigram probabilities
def score(m, probabilities):
    S = 0

    for i in range(len(m) - 1):
        ij = m[i] + m[i+1]
        S += math.log(probabilities[ij])

    return S
```

## Question 6 (Core): 

We start by writing a function that implements the Metropolis Algorithm. 

```python
all_char = list(string.ascii_uppercase) + [' ']

def metropolis_algo(m, T): 
    if T <= 0:
        return "Error: T is not greater than 0."
    # Keeping the old message
    old_message = m
    # Assigning two random characters to i and j
    i = all_char[random.randint(0, len(all_char) - 1)]
    j = all_char[random.randint(0, len(all_char) - 1)]
    # Checking i and j are not the same character
    while i == j: 
        j = all_char[random.randint(0, len(all_char) - 1)]

    # Swapping i and j
    new_message = swaproles(m,i,j)
    print(f"Score of old message: {score(old_message)} \nScore of new message: {score(new_message)}")
    # print(f"{i}, {j} \nOld message: {old_message} \nNew message: {new_message}")
    print(f"{i}, {j}, {new_message == old_message}")

    # If score of new message greater than score of old message, then we replace
    if score(new_message) > score(old_message):
        return new_message
    # Find the exponential expression value
    prob = math.e**((score(new_message) - score(old_message)) / T)
    # If randomly generated number b/w 0 and 1 is above our probability, will replace with new message
    if random.random() > prob:
        return new_message
    print("Not replaced.")
    return old_message
```

Above, I've written a test function for the metropolis algorithm that takes a message m and parameter T>0 and returns an updated message based on two randomly selected characters. Now, we will run this function multiple times. 

```python
n = 100
T = 1
m1 = ciphertexts[8]
print(f"Original message: \n{m1}")

for i in range(n):
    m1 = metropolis_algo(m1,T)
    if i % 1000 == 0: 
        print(f"Iteration {i}: {m1}")

print(f"Final result: {m1}")
```
